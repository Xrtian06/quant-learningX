# Matplotlib 绘制 K线图

> 日期: 2026-04-17

> 描述: 学习使用 matplotlib 和 mplfinance 绘制股票的 K线图


```python
import pandas as pd
import numpy as np

# 生成模拟 OHLC 数据
np.random.seed(42)
dates = pd.date_range('2025-01-01', periods=60, freq='B')
data = []
price = 100
for date in dates:
    open_p = price + np.random.randn() * 0.5
    close_p = open_p + np.random.randn() * 1.0
    high_p = max(open_p, close_p) + abs(np.random.randn()) * 0.5
    low_p = min(open_p, close_p) - abs(np.random.randn()) * 0.5
    data.append([date, open_p, high_p, low_p, close_p])
    price = close_p

df = pd.DataFrame(data, columns=['Date', 'Open', 'High', 'Low', 'Close'])
df.set_index('Date', inplace=True)
df.head()
```

```python
# 使用 mplfinance 绘制 K线图
import mplfinance as mpf

mpf.plot(df, type='candle', style='charles', title='Sample Candlestick Chart', ylabel='Price')
```

## 学习总结

1. `mplfinance` 是 matplotlib 的扩展包，专门用于金融数据可视化
2. 数据需要包含 `Open`, `High`, `Low`, `Close` 四列，且索引为日期
3. `type='candle'` 可以绘制标准的 K线图
4. 还可以叠加移动平均线：`mav=(5, 20, 60)`
